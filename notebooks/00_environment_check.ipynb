# 00 — Environment check

First milestone: verify a Colab GPU runtime. This notebook does not install dependencies, download data, or connect the runtime to the private repository.

Choose a GPU runtime in Colab, then run the cell below. Save a working copy to Drive if needed; local runtime files are temporary. Never paste passwords or access tokens into notebook cells.


```python
import sys
import platform
import subprocess
import torch

print('Python:', sys.version)
print('Platform:', platform.platform())
print('PyTorch:', torch.__version__)
print('PyTorch CUDA:', torch.version.cuda)
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
    print('GPU memory (GiB):', round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2))
else:
    print('No GPU detected. Select a GPU runtime in Colab and rerun this cell.')
try:
    result = subprocess.run(['nvidia-smi'], capture_output=True, text=True, check=False)
    print(result.stdout or result.stderr)
except FileNotFoundError:
    print('nvidia-smi is unavailable in this runtime.')
```
